## 4. Mine agent behavior

Once you have successfully placed an order, it's time to analyze the agents' behaviors and understand the underlying processes. The following code snippet reads the traces generated from your interactions with the coffee shop and creates a basic event log. This event log contains all your past interactions with the shop and can be utilized for process mining analysis.

One conversation can consist of multiple traces, as each trace comprises one input message provided by you and the subsequent actions taken by the agents. In other words, each message you send to the coffee shop creates a new trace.

```python
from src import TraceProcessor

processor = TraceProcessor()

# Export options:
# - export_as_json=True: Export as JSON instead of CSV
# - export_as_xes=True: Also generate XES format (standard for process mining tools)
processor.process_all_traces(export_as_json=False, export_as_xes=True)
```

## 5. Analyze with Process Mining Tools

The generated CSV/XES event log can now be imported into any process mining tool for analysis. The event log is located in the `generated_event_log/` folder.

### Recommended Tools

**Open Source Options:**
- [ProM](http://www.promtools.org/) - Comprehensive process mining framework
- [pm4py](https://pm4py.fit.fraunhofer.de/) - Python library for process mining

**Commercial Options:**
- Any commercial process mining platform that supports CSV/XES import

### Column Configuration

When importing the CSV file into your process mining tool, use the following column mappings:

**Required Columns:**
- **Case ID**: `case_id`
- **Activity Name**: `concept:instance`
- **Timestamp**: `time:timestamp`

**Additional Columns (for enhanced analysis):**
- `identity:id` → Text
- `concept:name` → Choice/Categorical
- `org:resource` → Choice/Categorical (agent name)
- `time_finished` → DateTime
- `duration` → Number/Numeric
- `model` → Choice/Categorical (LLM model)
- `input_tokens` → Number/Numeric
- `response_tokens` → Number/Numeric
- `message` → Text
- `tool` → Choice/Categorical

### What to Analyze

Once imported, you can perform various analyses:

1. **Process Discovery**: Visualize the agent interaction patterns and handoffs
2. **Performance Analysis**: Measure response times and identify bottlenecks
3. **Resource Analysis**: See which agents are most active
4. **Token Consumption**: Analyze LLM usage and costs
5. **Variant Analysis**: Discover different execution paths agents take
